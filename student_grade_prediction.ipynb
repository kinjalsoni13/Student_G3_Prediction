```python
X = df.drop('G3', axis=1)
y = df['G3']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
categorical_cols = X.select_dtypes(include=['object']).columns

print("Categorical columns:")
print(categorical_cols)
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print("Original shape:", X.shape)
print("Encoded shape:", X_encoded.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Actual values:")
print(y_test.values[:10])

print("\nPredicted values:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest MAE:", rf_mae)
print("Random Forest RMSE:", rf_rmse)
print("Random Forest R² Score:", rf_r2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(y_test, rf_pred)

plt.xlabel("Actual G3")
plt.ylabel("Predicted G3")
plt.title("Actual vs Predicted G3 - Random Forest")

plt.show()
```

```python
feature_importance = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

print(feature_importance.head(10))
```

```python
X_no_grades = X_encoded.drop(['G1', 'G2'], axis=1)

X_train_ng, X_test_ng, y_train_ng, y_test_ng = train_test_split(
    X_no_grades,
    y,
    test_size=0.2,
    random_state=42
)

print("X without G1/G2:", X_no_grades.shape)
print("Training:", X_train_ng.shape)
print("Testing:", X_test_ng.shape)
```

```python
rf_model_ng = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf_model_ng.fit(X_train_ng, y_train_ng)

rf_pred_ng = rf_model_ng.predict(X_test_ng)

mae_ng = mean_absolute_error(y_test_ng, rf_pred_ng)
rmse_ng = np.sqrt(mean_squared_error(y_test_ng, rf_pred_ng))
r2_ng = r2_score(y_test_ng, rf_pred_ng)

print("MAE:", mae_ng)
print("RMSE:", rmse_ng)
print("R² Score:", r2_ng)
```

```python
from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    rf_model,
    X_encoded,
    y,
    cv=5,
    scoring='r2'
)

print("R² scores:", cv_scores)
print("Mean R²:", cv_scores.mean())
print("Std R²:", cv_scores.std())
```
